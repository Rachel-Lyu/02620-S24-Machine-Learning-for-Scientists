```python
import pandas as pd

# Define the path to the file
file_path = '/content/drive/My Drive/Spellman.csv'  # Adjust the path accordingly

# Read the CSV file into a Pandas DataFrame
df = pd.read_csv(file_path)

# Display the DataFrame
print(df)
```

```python
import matplotlib.pyplot as plt
```

```python

# Choose the gene to plot
gene = 'YAL001C'

# Extract the data for that gene
gene_data = df[df['time'] == gene].iloc[0, 1:]

# Extract time points from columns
time_points = list(map(int, gene_data.index.tolist()))

# Convert gene expression data to floats
gene_expression = gene_data.astype(float).tolist()

# Create the plot
plt.figure(figsize=(10, 5))
plt.plot(time_points, gene_expression, marker='o')
plt.title(f'Expression Levels of {gene} Over Time')
plt.xlabel('Time')
plt.ylabel('Expression Level')
plt.grid(True)
plt.show()
```

```python
# Extract initial expression levels from the column corresponding to time=40
initial_expression = df['40'].astype(float)

# Create the histogram
plt.figure(figsize=(10, 5))
plt.hist(initial_expression, bins=20, edgecolor='black')
plt.title('Histogram of Initial Expression Levels of All Genes')
plt.xlabel('Initial Expression Level')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
# Extract initial expression levels from the column corresponding to time=40
final_expression = df['260'].astype(float)

# Create the histogram
plt.figure(figsize=(10, 5))
plt.hist(final_expression, bins=20, edgecolor='black')
plt.title('Histogram of Final Expression Levels of All Genes')
plt.xlabel('Final Expression Level')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
# Create the histogram
plt.figure(figsize=(10, 5))

# Plot initial expression levels
plt.hist(initial_expression, bins=20, alpha=0.5, edgecolor='black', label='Initial (time=40)')

# Plot final expression levels
plt.hist(final_expression, bins=20, alpha=0.5, edgecolor='black', label='Final (time=260)')

plt.title('Histogram of Initial and Final Expression Levels of All Genes')
plt.xlabel('Expression Level')
plt.ylabel('Frequency')
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```

```python
# Create the figure and subplots
plt.figure(figsize=(10, 10))

# Plot initial expression levels
plt.subplot(2, 1, 1)  # 2 rows, 1 column, first plot
plt.hist(initial_expression, bins=20, edgecolor='black')
plt.title('Histogram of Initial Expression Levels (time=40)')
plt.xlabel('Expression Level')
plt.ylabel('Frequency')
plt.xlim([-3,3])
plt.grid(True)

# Plot final expression levels
plt.subplot(2, 1, 2)  # 2 rows, 1 column, second plot
plt.hist(final_expression, bins=20, edgecolor='black')
plt.title('Histogram of Final Expression Levels (time=260)')
plt.xlabel('Expression Level')
plt.ylabel('Frequency')
plt.xlim([-3,3])
plt.grid(True)

# Add space between subplots for better readability
plt.tight_layout()

# Show the plots
plt.show()
```

```python
import csv

# Define the path to the file
file_path = '/content/drive/My Drive/Spellman.csv'  # Adjust the path accordingly

# Initialize a dictionary to hold the data and list to hold header (time points)
data = {}
header = []

# Read the file
with open(file_path, 'r') as f:
    csv_reader = csv.reader(f)
    header = next(csv_reader)  # Read and store the header
    for row in csv_reader:
        gene = row[0]
        expression_levels = [float(x) for x in row[1:]]
        data[gene] = expression_levels
print(data)
```

```python
import matplotlib.pyplot as plt

# Choose a specific gene
specific_gene = 'YAL001C'

# Retrieve the data for the specific gene
expression_levels = data.get(specific_gene, [])

# Extract time points for plotting (omitting the 'time' from the header)
time_points = header[1:]

# Create a plot
plt.figure(figsize=(10, 5))

# Plot the expression levels for the specific gene
plt.plot(time_points, expression_levels, marker='o')

plt.title('Expression Levels of ' + specific_gene + ' Over Time')
plt.xlabel('Time Points')
plt.ylabel('Expression Level')
plt.grid(True)

plt.show()

```

```python

# Initialize a list to store initial expression levels of all genes
initial_expression_levels = []

# Iterate through the dictionary to collect initial expression levels
for gene, expression_levels in data.items():
    if expression_levels:  # Check if the list is not empty
        initial_expression_levels.append(expression_levels[0])

# Create a histogram
plt.figure(figsize=(10, 5))

# Plot the histogram
plt.hist(initial_expression_levels, bins=50, edgecolor='black')

# Add titles and labels
plt.title('Histogram of Initial Expression Levels of All Genes')
plt.xlabel('Initial Expression Level')
plt.ylabel('Frequency')

# Add a grid
plt.grid(True)

# Show the plot
plt.show()

```

```python

# Initialize a list to store final expression levels of all genes
final_expression_levels = []

# Iterate through the dictionary to collect final expression levels
for gene, expression_levels in data.items():
    if expression_levels:  # Check if the list is not empty
        final_expression_levels.append(expression_levels[-1])  # Take the last element as the final expression level

# Create a histogram
plt.figure(figsize=(10, 5))

# Plot the histogram
plt.hist(final_expression_levels, bins=50, edgecolor='black')

# Add titles and labels
plt.title('Histogram of Final Expression Levels of All Genes')
plt.xlabel('Final Expression Level')
plt.ylabel('Frequency')

# Add a grid
plt.grid(True)

# Show the plot
plt.show()
```
