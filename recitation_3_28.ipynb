```python
def viterbi(obs, states, start_prob, trans_prob, emit_prob):
    V = np.zeros((len(obs), len(states)))  # Initialize the Viterbi matrix
    path = np.zeros((len(obs), len(states)), dtype=int)  # Initialize the path matrix

    # Initial population of the Viterbi matrix and path matrix
    for s in range(len(states)):
        V[0][s] = start_prob[s] * emit_prob[s][obs[0]]
        path[0][s] = s

    # Viterbi algorithm
    for t in range(1, len(obs)):
        new_path = np.zeros((len(states), len(states)), dtype=int)
        for y in range(len(states)):
            (prob, state) = max((V[t-1][y0] * trans_prob[y0][y] * emit_prob[y][obs[t]], y0) for y0 in range(len(states)))
            V[t][y] = prob
            new_path[y][:] = path[t-1][state]
            new_path[y][t] = y
        path = new_path

    # Find the most probable last state
    n = len(obs) - 1  # Index of the last observation
    (prob, state) = max((V[n][y], y) for y in range(len(states)))

    return path[state], prob

# Parameters as per your problem
states = [0, 1, 2, 3]  # Corresponding to P, E, R, B
observations = [0, 1]  # 0 for non-methylated, 1 for methylated

start_probability = np.array([0.2, 0.3, 0.3, 0.2])
transition_probability = np.array([
    [0.3, 0.1, 0.3, 0.3],
    [0.1, 0.4, 0.1, 0.4],
    [0.1, 0.1, 0.4, 0.4],
    [0.1, 0.1, 0.1, 0.7]
])
emission_probability = np.array([
    [0.9, 0.1],  # P
    [0.8, 0.2],  # E
    [0.9, 0.1],  # R
    [0.2, 0.8]   # B
])

# Observation sequence
obs_seq = np.array([1, 0, 1, 1, 0])  # Replace with your actual observations

# Apply the Viterbi algorithm
predicted_states, max_prob = viterbi(obs_seq, states, start_probability, transition_probability, emission_probability)

# Translate states back to labels if needed
state_labels = ['P', 'E', 'R', 'B']
predicted_state_sequence = [state_labels[state] for state in predicted_states]

print("Most likely states:", predicted_state_sequence)
print("Probability of the state sequence:", max_prob)
```
